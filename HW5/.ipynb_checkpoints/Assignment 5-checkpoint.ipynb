## Question 2

*Notes, Chapter 12, Exercise 6.*

Implement a two layer network in JAX and verify `jax.grad` against a finite difference gradient to five digits. State which precision you used and why it matters.

```python
# Question 2: your code here

def init_params(key, d_in=3, width=8, d_out=1):
    k1, k2, k3, k4 = jax.random.split(key, 4)
    return {
        "W1": jax.random.normal(k1, (width, d_in)) / jnp.sqrt(d_in),
        "b1": jax.random.normal(k2, (width,)),
        "W2": jax.random.normal(k3, (d_out, width)) / jnp.sqrt(width),
        "b2": jax.random.normal(k4, (d_out,)),
    }

def forward(p, X):
    # X: (n, d_in). Layer 1: sigmoid hidden layer; layer 2: linear output
    h = jax.nn.sigmoid(X @ p["W1"].T + p["b1"])
    return h @ p["W2"].T + p["b2"]

def loss(p, X, y):
    return 0.5 * jnp.mean((forward(p, X)[:, 0] - y) ** 2)

def fd_grad(f, p, eps):
    g = {}
    for name, val in p.items():
        flat = np.array(val, dtype=val.dtype).ravel()
        gflat = np.zeros_like(flat)
        for j in range(flat.size):
            e = np.zeros_like(flat); e[j] = eps
            pp = dict(p); pp[name] = jnp.asarray((flat + e).reshape(val.shape))
            pm = dict(p); pm[name] = jnp.asarray((flat - e).reshape(val.shape))
            gflat[j] = (f(pp) - f(pm)) / (2 * eps)
        g[name] = gflat.reshape(val.shape)
    return g

def compare(dtype, eps):
    key = jax.random.PRNGKey(0)
    kp, kx, ky = jax.random.split(key, 3)
    p = jax.tree_util.tree_map(lambda a: a.astype(dtype), init_params(kp))
    X = jax.random.normal(kx, (50, 3)).astype(dtype)
    y = jax.random.normal(ky, (50,)).astype(dtype)
    f = lambda q: loss(q, X, y)
    g_ad = jax.grad(f)(p)
    g_fd = fd_grad(f, p, eps)
    ad = np.concatenate([np.ravel(g_ad[k]) for k in p])
    fd = np.concatenate([np.ravel(g_fd[k]) for k in p])
    rel = np.linalg.norm(ad - fd) / np.linalg.norm(ad)
    print(f"{str(np.dtype(dtype)):8s} eps={eps:.0e}  relative error = {rel:.2e}"
          f"  -> agreeing digits ~ {-np.log10(rel):.1f}")
    return ad, fd

ad, fd = compare(jnp.float64, 1e-6)
print("first 5 entries, jax.grad :", np.round(ad[:5], 8))
print("first 5 entries, FD       :", np.round(fd[:5], 8))
compare(jnp.float32, 1e-3)   # same check in single precision
compare(jnp.float32, 1e-6)   # single precision with a 'double-precision' step
```

## Question 6

*Notes, Chapter 12, Exercise 21.*

Reproduce Figure 12.2. Build 14 layer networks of width 32 with sigmoid, ReLU, and residual layers, and at initialization measure $\|\partial J / \partial \mathbf{b}^\ell\|$ for each $\ell$, averaged over 120 draws. Plot on a log scale, fit the sigmoid slope, and compare with $-\log_{10}(0.25)$.

```python
# Question 6: your code here
import matplotlib.pyplot as plt

L, WIDTH, DRAWS = 14, 32, 120

def init(key):
    keys = jax.random.split(key, 2 * L)
    Ws = jnp.stack([jax.random.normal(keys[2*i], (WIDTH, WIDTH)) * jnp.sqrt(2.0 / WIDTH)
                    for i in range(L)])
    bs = jnp.stack([jax.random.normal(keys[2*i+1], (WIDTH,)) for i in range(L)])
    return Ws, bs

def make_loss(kind):
    act = jax.nn.relu if kind == "relu" else jax.nn.sigmoid
    def J(bs, Ws, x, target):
        y = x
        for l in range(L):
            z = Ws[l] @ y + bs[l]
            y = y + act(z) if kind == "resnet" else act(z)
        return 0.5 * jnp.sum((y - target) ** 2)
    # gradient with respect to all biases at once: shape (L, WIDTH)
    return jax.jit(jax.grad(J))

def layer_grad_norms(kind, seed=0):
    gJ = make_loss(kind)
    norms = np.zeros((DRAWS, L))
    key = jax.random.PRNGKey(seed)
    for d in range(DRAWS):
        key, kp, kx, kt = jax.random.split(key, 4)
        Ws, bs = init(kp)
        x = jax.random.normal(kx, (WIDTH,))
        t = jax.random.normal(kt, (WIDTH,))
        g = gJ(bs, Ws, x, t)
        norms[d] = np.linalg.norm(np.asarray(g), axis=1)   # ||dJ/db^l|| per layer
    return norms.mean(axis=0)                               # average over draws

layers = np.arange(1, L + 1)
results = {k: layer_grad_norms(k) for k in ["sigmoid", "relu", "resnet"]}

# fit log10 ||dJ/db^l|| = a + s * l for the sigmoid curve
s, a = np.polyfit(layers, np.log10(results["sigmoid"]), 1)
print(f"sigmoid slope           : {s:.3f} decades per layer")
print(f"prediction -log10(0.25) : {-np.log10(0.25):.3f}")
print(f"layer 1 / layer 14 (sigmoid): 10^{np.log10(results['sigmoid'][0] / results['sigmoid'][-1]):.2f}")
for k, v in results.items():
    print(f"{k:8s} log10||dJ/db^l||, l=1..14:", np.round(np.log10(v), 2))

plt.figure(figsize=(6, 4))
style = {"sigmoid": "tab:red", "relu": "navy", "resnet": "teal"}
for k, v in results.items():
    plt.plot(layers, np.log10(v), "o-", color=style[k], label=k)
plt.plot(layers, a + s * layers, "k--", lw=1, label=f"sigmoid fit, slope {s:.3f}")
plt.xlabel(r"layer $\ell$"); plt.ylabel(r"$\log_{10}\|\partial J/\partial \mathbf{b}^\ell\|$")
plt.title("Gradient magnitude at initialization (120 draws)")
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout()
plt.savefig("HW5Problem6.jpg")
```

## Question 8

*Notes, Chapter 13, Exercise 7.*

Reproduce Example 13.3.1. Build snapshots as random combinations of $\sin(k\pi x)$ on $d = 200$ points with amplitudes falling like $k^{-3/2}$, add a little sensor noise, and solve (13.2). Confirm the cumulative energies 83.1%, 94.2%, 97.0% and the inner products against the true modes. Then break it: give two modes nearly equal amplitudes and report what happens to the corresponding principal directions. Why does the method stop recovering them individually?

```python
# Question 8: your code here
import matplotlib.pyplot as plt

d, n, K = 200, 500, 10           # grid points, snapshots, number of sine modes
noise = 1e-3                     # sensor-noise standard deviation
x = np.linspace(0, 1, d)
modes = np.stack([np.sin(k * np.pi * x) for k in range(1, K + 1)], axis=1)   # (d, K)
modes /= np.linalg.norm(modes, axis=0)                                       # unit-norm true modes

def snapshots(amps, seed=0):
    rng = np.random.default_rng(seed)
    C = rng.standard_normal((K, n)) * amps[:, None]       # random coefficients
    X = modes @ C + noise * rng.standard_normal((d, n))   # add sensor noise
    return X - X.mean(axis=1, keepdims=True)              # center the data

def pca(X):
    # (13.2): X X^T Xhat = Xhat Lambda -> eigen-decomposition of the d x d matrix X X^T
    lam, V = np.linalg.eigh(X @ X.T)
    idx = np.argsort(lam)[::-1]
    return lam[idx], V[:, idx]

def report(amps, label, seed=0):
    X = snapshots(amps, seed)
    lam, V = pca(X)
    cum = np.cumsum(lam) / lam.sum()
    print(f"--- {label} ---")
    print("cumulative energy r=1..5:", np.round(100 * cum[:5], 1), "%")
    G = np.abs(V[:, :4].T @ modes[:, :4])                  # |<principal dir i, true mode k>|
    print("|<v_i, sin(k pi x)>|, rows i=1..4, cols k=1..4:\n", np.round(G, 3))
    return lam, V

amps = np.arange(1, K + 1) ** -1.5
lam, V = report(amps, "amplitudes k^(-3/2)")

# Break it: give modes 2 and 3 nearly equal amplitudes
amps_b = amps.copy(); amps_b[2] = amps_b[1] * 0.98
lam_b, V_b = report(amps_b, "modes 2 and 3 with nearly equal amplitudes")
print("eigenvalue gap lambda2-lambda3 (relative):", (lam_b[1] - lam_b[2]) / lam_b[1])

# Repeat the broken case over several random draws: v2, v3 are rotated mixtures of modes 2 and 3
for seed in range(4):
    X = snapshots(amps_b, seed); _, Vs = pca(X)
    c = Vs[:, 1] @ modes[:, 1:3]
    print(f"draw {seed}: v2 = {c[0]:+.3f} sin(2 pi x) {c[1]:+.3f} sin(3 pi x)")

fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
for i in range(3):
    s = np.sign(V[:, i] @ modes[:, i])
    ax[0].plot(x, s * V[:, i], label=f"$v_{i+1}$")
    ax[1].plot(x, V_b[:, i], label=f"$v_{i+1}$")
ax[0].set_title("Amplitudes $k^{-3/2}$: modes recovered")
ax[1].set_title("Modes 2, 3 nearly equal: $v_2, v_3$ mixed")
for a in ax: a.legend(); a.set_xlabel("x")
plt.tight_layout(); plt.show()
```

## Question 10

*Notes, Chapter 13, Exercise 13.*

Show that the reconstruction error equals the sum of the discarded eigenvalues, $\sum_{i>r} \lambda_i$, and hence that PCA minimizes it among all rank $r$ projections (Eckart–Young). Check the claim numerically against Figure 13.6a.

```python
# Question 10: your code here
import matplotlib.pyplot as plt

# Model problem of Definition 13.8.1, discretized by (13.5). Reused in Questions 11 and 12.
c, nu = 0.9, 0.004
nx = 200
dx = 1.0 / (nx - 1)
dt = 2e-4
nsteps, every = 1400, 20
xg = np.linspace(0, 1, nx)

def initial(width=0.055):
    u = np.exp(-((xg - 0.18) / width) ** 2)
    u[0] = u[-1] = 0.0
    return u

def step(u):
    un = u.copy()
    un[1:-1] = u[1:-1] + dt * (nu * (u[2:] - 2 * u[1:-1] + u[:-2]) / dx**2
                               - c * (u[2:] - u[:-2]) / (2 * dx))
    return un

def snapshot_matrix(width=0.055):
    u, cols = initial(width), []
    for k in range(nsteps):
        if k % every == 0:
            cols.append(u.copy())
        u = step(u)
    return np.stack(cols, axis=1)

X = snapshot_matrix()
print("snapshot matrix shape:", X.shape)

# Eigenvalues of X X^T are the squared singular values of X
U, S, VT = np.linalg.svd(X, full_matrices=False)
lam = S ** 2

print(" r | ||(I - U_r U_r^T) X||_F^2 | sum_{i>r} lambda_i | relative difference")
for r in [1, 2, 4, 8, 12]:
    Ur = U[:, :r]
    err = np.linalg.norm(X - Ur @ (Ur.T @ X), "fro") ** 2
    tail = lam[r:].sum()
    print(f"{r:2d} | {err:24.6e} | {tail:18.6e} | {abs(err - tail) / tail:.1e}")

# Eckart-Young: no other rank-r orthogonal projection does better.
rng = np.random.default_rng(0)
r = 4
pca_err = np.linalg.norm(X - U[:, :r] @ (U[:, :r].T @ X), "fro") ** 2
rand_errs = []
for _ in range(2000):
    # random r-dimensional subspace, biased toward the PCA one so the competitors are close
    Q, _ = np.linalg.qr(U[:, :r] + 0.01 * rng.standard_normal((nx, r)))
    rand_errs.append(np.linalg.norm(X - Q @ (Q.T @ X), "fro") ** 2)
print(f"\nr = 4: PCA error {pca_err:.4e}; best of 2000 perturbed subspaces {min(rand_errs):.4e}")
print("PCA beats every perturbed subspace:", pca_err <= min(rand_errs))

# Compare with Figure 13.6a: fraction of energy left out = sum_{i>r} lambda_i / sum_i lambda_i
energy_left = 1 - np.cumsum(lam) / lam.sum()
print("energy left out after r = 1..8 modes:", np.round(energy_left[:8], 4))
print(f"energy captured by r = 4: {100 * (1 - energy_left[3]):.1f}%  (Figure 13.6a: 99.3%)")

errs = [np.linalg.norm(X - U[:, :r] @ (U[:, :r].T @ X), "fro") ** 2 for r in range(1, 21)]
plt.figure(figsize=(5.5, 3.8))
plt.semilogy(range(1, 21), errs, "o", label=r"$\|(I-U_rU_r^T)X\|_F^2$")
plt.semilogy(range(1, 21), [lam[r:].sum() for r in range(1, 21)], "-", label=r"$\sum_{i>r}\lambda_i$")
plt.xlabel("r"); plt.legend(); plt.grid(alpha=0.3); plt.title("Reconstruction error vs discarded eigenvalues")
plt.tight_layout(); plt.show()
```

## Question 12

*Notes, Chapter 13, Exercise 23.*

Build the Galerkin reduced order model of Section 13.6 for Definition 13.8.1. Write (13.5) as $\frac{d\mathbf{x}}{dt} = \mathbf{A}\mathbf{x}$ by assembling the $200 \times 200$ matrix $\mathbf{A}$ explicitly, project it to $\mathbf{A}_r = \mathbf{V}^T \mathbf{A} \mathbf{V}$, and march the reduced system with the same $\Delta t$. Compare reduced and full solutions for $r = 2, 4, 8$, plot error against $r$ alongside the discarded singular value energy, and report the speedup. Then change the initial pulse width from 0.055 to 0.11, a state not represented among the snapshots, and comment on what happens to the reduced model.

```python
# Question 12: your code here
import time
import matplotlib.pyplot as plt

# 1. Assemble A explicitly so that (13.5) is x^{k+1} = x^k + dt * A x^k.
#    Rows 0 and nx-1 stay zero, which keeps the Dirichlet values u_0 = u_199 = 0 fixed.
alpha = nu / dx**2 + c / (2 * dx)
beta  = -2 * nu / dx**2
gamma = nu / dx**2 - c / (2 * dx)
A = np.zeros((nx, nx))
for i in range(1, nx - 1):
    A[i, i - 1], A[i, i], A[i, i + 1] = alpha, beta, gamma
print(f"alpha = {alpha:.2f}, beta = {beta:.2f}, gamma = {gamma:.2f}")

# check: one step with A matches the stencil update of (13.5)
u0 = initial()
print("max |(I + dt A) u0 - step(u0)| =", np.abs(u0 + dt * A @ u0 - step(u0)).max())

def march_full(u0, A, n=nsteps):
    traj = np.empty((n + 1, u0.size)); traj[0] = u = u0
    for k in range(n):
        u = u + dt * (A @ u)
        traj[k + 1] = u
    return traj

def march_reduced(u0, Ar, V, n=nsteps):
    a = V.T @ u0
    traj = np.empty((n + 1, a.size)); traj[0] = a
    for k in range(n):
        a = a + dt * (Ar @ a)
        traj[k + 1] = a
    return traj

# 2. POD basis from the training snapshots (pulse width 0.055)
X = snapshot_matrix(0.055)
U, S, _ = np.linalg.svd(X, full_matrices=False)
discarded = lambda r: np.sum(S[r:]**2) / np.sum(S**2)

def compare(width, rs=(1, 2, 3, 4, 6, 8, 10, 12), label=""):
    u0 = initial(width)
    full = march_full(u0, A)
    errs = {}
    for r in rs:
        V = U[:, :r]
        Ar = V.T @ A @ V
        red = march_reduced(u0, Ar, V) @ V.T          # lift back to 200 dimensions
        errs[r] = np.linalg.norm(red - full) / np.linalg.norm(full)   # error over all time steps
    print(f"\n{label}")
    print(" r | rel. error (all t) | discarded energy sum_{i>r} s_i^2 / sum s_i^2")
    for r in rs:
        print(f"{r:2d} | {errs[r]:18.3e} | {discarded(r):.3e}")
    return full, errs

full, errs = compare(0.055, label="Training initial condition (width 0.055)")

# 3. Speedup (online cost only: marching 1400 steps)
def best_time(f, reps=5):
    ts = []
    for _ in range(reps):
        t0 = time.perf_counter(); f(); ts.append(time.perf_counter() - t0)
    return min(ts)
t_full = best_time(lambda: march_full(u0, A))
print(f"\nfull model (200 x 200), 1400 steps: {1e3 * t_full:.1f} ms")
for r in (2, 4, 8):
    V = U[:, :r]; Ar = V.T @ A @ V
    t_red = best_time(lambda: march_reduced(u0, Ar, V))
    print(f"reduced r = {r}: {1e3 * t_red:.1f} ms  -> speedup {t_full / t_red:.1f}x"
          f"  (flop ratio per step: {nx**2 / r**2:.0f}x)")

# 4. Plots: final-time profiles and error against r
fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].plot(xg, full[-1], "k", lw=2.5, label="full (200)")
for r, col in [(2, "tab:red"), (4, "tab:orange"), (8, "tab:blue")]:
    V = U[:, :r]
    red = march_reduced(initial(), V.T @ A @ V, V)[-1] @ V.T
    ax[0].plot(xg, red, "--", color=col, label=f"ROM r = {r}")
ax[0].set_title("Solution at T = 0.28"); ax[0].set_xlabel("x"); ax[0].legend()
rs = list(errs)
ax[1].semilogy(rs, [errs[r] for r in rs], "o-", label="ROM relative error")
ax[1].semilogy(rs, [discarded(r) for r in rs], "s--", label="discarded energy")
ax[1].semilogy(rs, [np.sqrt(discarded(r)) for r in rs], "^:", label="sqrt(discarded energy)")
ax[1].set_xlabel("r"); ax[1].legend(); ax[1].grid(alpha=0.3); ax[1].set_title("Error vs r")
plt.tight_layout(); plt.show()

# 5. Out-of-sample initial condition: pulse width 0.11, same basis
full_w, errs_w = compare(0.11, label="New initial condition (width 0.11), basis from width 0.055")
u0w = initial(0.11)
print(f"\nprojection error of the new initial condition onto span(V):")
for r in (2, 4, 8):
    V = U[:, :r]
    print(f"  r = {r}: ||u0 - V V^T u0|| / ||u0|| = {np.linalg.norm(u0w - V @ (V.T @ u0w)) / np.linalg.norm(u0w):.3f}")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
for a, k, title in [(ax[0], 0, "t = 0"), (ax[1], nsteps, "t = 0.28")]:
    a.plot(xg, full_w[k], "k", lw=2.5, label="full (200)")
    for r, col in [(4, "tab:orange"), (8, "tab:blue")]:
        V = U[:, :r]
        a.plot(xg, march_reduced(u0w, V.T @ A @ V, V)[k] @ V.T, "--", color=col, label=f"ROM r = {r}")
    a.set_title(f"Width 0.11, {title}"); a.set_xlabel("x"); a.legend()
plt.tight_layout(); plt.show()
```
